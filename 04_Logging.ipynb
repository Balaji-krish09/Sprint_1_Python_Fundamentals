## Logging: 1. What is Logging?

### Explanation
Logging is the practice of recording messages about a program's 
execution - like what happened, when, and how severe it was - to 
a file, console, or other output, instead of just showing 
something once on screen and losing it.

```python
import logging

logging.warning("This is a log message!")

# Where it's used: 'logging.warning(...)' generates a log record.
# Unlike print(), this message carries a severity level (WARNING)
# and can be routed to files, timestamped, and filtered - not just
# displayed once.
```

## Logging: 2. Why use Logging?

### Explanation
Logging helps track what a program is doing over time, diagnose 
issues after they've happened (not just while watching the screen 
live), and keep a permanent record - especially useful for 
programs that run unattended, like servers or scheduled scripts.

```python
import logging

logging.basicConfig(level=logging.INFO)
logging.info("Program started successfully.")
logging.error("Something went wrong while processing data.")

# Where it's used: both log lines get automatically tagged with
# their severity (INFO, ERROR). Unlike print(), you could later
# filter to see ONLY errors, or send all of this to a file for
# permanent record-keeping - useful for tracking a program's
# behavior over time, not just watching it live.
```

## Logging: 3. print() vs logging

### Explanation
print() only displays text on screen, temporarily, with no severity 
level, no timestamp, and no way to control what's shown. logging 
supports severity levels, timestamps, multiple output destinations 
(file/console), and can be turned on/off without editing the code 
itself.

```python
import logging
logging.basicConfig(level=logging.DEBUG, format="%(asctime)s - %(levelname)s - %(message)s")

print("This is just plain text, no context.")
logging.debug("This has a timestamp, severity level, and can be filtered.")

# Where it's used: print() gives plain unstructured text. The
# logging.debug() line, because of basicConfig's format string,
# automatically adds a timestamp and severity level (DEBUG) - 
# information print() has no way of attaching.
```

## Logging: 4. Log Levels

### Explanation
Log levels indicate the SEVERITY of a message, in increasing order:
DEBUG (detailed diagnostic info) -> INFO (general confirmation 
things are working) -> WARNING (something unexpected, but not 
breaking) -> ERROR (a real problem occurred) -> CRITICAL (a severe 
error, program may not continue). You can filter logs to show only 
a certain severity and above.

```python
import logging
logging.basicConfig(level=logging.DEBUG, format="%(levelname)s - %(message)s")

logging.debug("Debugging details for developers.")
logging.info("Program is running normally.")
logging.warning("Something unexpected happened, but it's not breaking.")
logging.error("A real error occurred.")
logging.critical("Critical failure - program may not continue!")

# Where it's used: each line calls a DIFFERENT level method 
# (debug, info, warning, error, critical). Because level=logging.DEBUG
# was set in basicConfig, ALL five levels are shown - if it had been
# set to level=logging.WARNING instead, only warning/error/critical
# would appear, since DEBUG and INFO are less severe.
```

## Logging: 5. Log Formatters

### Explanation
A formatter controls the LAYOUT of each log message - what 
information appears (timestamp, level, message, etc.) and in what 
order. Without a formatter, logs just show the bare message with 
no context.

```python
import logging

logger = logging.getLogger("formatter_demo")
handler = logging.StreamHandler()
formatter = logging.Formatter("%(asctime)s | %(levelname)s | %(message)s")
handler.setFormatter(formatter)
logger.addHandler(handler)
logger.setLevel(logging.INFO)

logger.info("This log now has a timestamp and level shown clearly.")

# Where it's used: 'logging.Formatter("%(asctime)s | %(levelname)s | %(message)s")'
# defines the LAYOUT - asctime (timestamp), levelname (severity), 
# and message, separated by "|". 'handler.setFormatter(formatter)' 
# attaches that layout to the handler, so every log line printed 
# through it follows this exact structured format.
```

## Logging: 6. File Handlers

### Explanation
A File Handler sends log messages to a FILE on disk, instead of 
just the console - so logs are saved permanently and can be 
reviewed later, even after the program has stopped running.

```python
import logging

logger = logging.getLogger("file_demo")
file_handler = logging.FileHandler("app.log")
formatter = logging.Formatter("%(asctime)s - %(levelname)s - %(message)s")
file_handler.setFormatter(formatter)
logger.addHandler(file_handler)
logger.setLevel(logging.INFO)

logger.info("This message is saved to app.log, not just shown on screen.")

# Where it's used: 'logging.FileHandler("app.log")' creates a
# handler that writes to a file named app.log. 'logger.addHandler(file_handler)'
# attaches it, so every log call on 'logger' now gets written to
# that file on disk - check your project folder, app.log should
# appear there after running this.
```

## Logging: 7. Console Handlers

### Explanation
A Console (Stream) Handler sends log messages to the terminal/
console output - the default behavior most people expect, useful 
for real-time monitoring while a program runs.

```python
import logging

logger = logging.getLogger("console_demo")
console_handler = logging.StreamHandler()
formatter = logging.Formatter("%(levelname)s - %(message)s")
console_handler.setFormatter(formatter)
logger.addHandler(console_handler)
logger.setLevel(logging.INFO)

logger.info("This message shows up directly in the console/terminal.")

# Where it's used: 'logging.StreamHandler()' creates a handler that
# outputs to the console (by default, sys.stderr). Unlike the File
# Handler above, nothing gets saved to disk here - it's purely for
# watching logs live while the program runs.
```

## Logging: 8 (Bonus). Rotating Logs

### Explanation
A Rotating File Handler automatically creates a NEW log file once 
the current one reaches a certain size, and keeps a limited number 
of old backup files - preventing a single log file from growing 
forever and eating up disk space.

```python
import logging
from logging.handlers import RotatingFileHandler

logger = logging.getLogger("rotating_demo")
rotating_handler = RotatingFileHandler("rotating_app.log", maxBytes=200, backupCount=3)
formatter = logging.Formatter("%(asctime)s - %(message)s")
rotating_handler.setFormatter(formatter)
logger.addHandler(rotating_handler)
logger.setLevel(logging.INFO)

for i in range(20):
    logger.info(f"Log message number {i}")

# Where it's used: 'maxBytes=200' means once rotating_app.log hits
# 200 bytes, it automatically renames it and starts a fresh file.
# 'backupCount=3' keeps only the last 3 old files (rotating_app.log.1,
# .2, .3), deleting older ones - check your folder, you should see
# multiple rotating_app.log.X files after running this.
```

## Logging: 9. Logging Best Practices

### Explanation
General guidelines for using logging effectively in real projects:
use appropriate log levels (don't log everything as ERROR), avoid 
logging sensitive data (passwords, personal info), use descriptive 
messages, configure logging once at the start of a program (not 
repeatedly), and prefer file/rotating logs over print() for 
anything meant to run unattended.

```python
import logging

# Good practice: configure once, at the top of your program
logging.basicConfig(
    level=logging.INFO,
    format="%(asctime)s - %(levelname)s - %(message)s",
    filename="best_practice_demo.log"
)

logging.info("User login successful.")           # good - clear, appropriate level
logging.warning("API response slower than usual.") # good - warning, not error
logging.error("Failed to connect to database.")    # good - real problem, correctly leveled

# Where it's used: this demonstrates the practices themselves -
# logging.basicConfig() is called ONCE, each message uses the
# level that actually matches its severity (not everything dumped
# as one level), and messages are clear/descriptive rather than
# vague like "something happened".
```
