## File Handling: 1. Reading Text Files

### Explanation
Reading a text file means opening it and retrieving its stored 
content into your program, so you can use or process that data.

```python
# First, create a sample file to read from
with open("sample.txt", "w") as f:
    f.write("Hello Balaji\nWelcome to Sprint 1\nFile Handling is fun!")

# Now read it back
with open("sample.txt", "r") as f:
    content = f.read()
print(content)

# Where it's used: open("sample.txt", "r") opens the file in READ
# mode ("r"). f.read() pulls the entire file's content into the
# variable 'content', which is then printed - this is the actual
# reading operation.
```

## File Handling: 2. Writing Text Files

### Explanation
Writing to a text file means saving data from your program into a 
file on disk, so it persists even after the program stops running. 
"w" mode overwrites the file completely; "a" mode appends without 
erasing existing content.

```python
# "w" mode - overwrites the file completely
with open("notes.txt", "w") as f:
    f.write("This is the first line.\n")

# "a" mode - appends without erasing what's already there
with open("notes.txt", "a") as f:
    f.write("This line was appended.\n")

with open("notes.txt", "r") as f:
    print(f.read())

# Where it's used: the first 'open("notes.txt", "w")' creates/
# overwrites the file with one line. The second 'open("notes.txt", "a")'
# adds a new line WITHOUT deleting the first one - the final read
# confirms both lines exist together.
```

## File Handling: 3. CSV Files

### Explanation
CSV (Comma-Separated Values) files store tabular data as plain 
text, where each line is a row and values are separated by commas. 
Python's csv module lets you read/write this structured data 
easily, instead of manually splitting strings by commas.

```python
import csv

# Writing to a CSV file
with open("students.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Name", "Age", "Course"])
    writer.writerow(["Balaji", 24, "AI/ML"])
    writer.writerow(["Priya", 22, "Web Dev"])

# Reading from a CSV file
with open("students.csv", "r") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)

# Where it's used: csv.writer(f).writerow([...]) writes each list
# as one comma-separated row in the file. csv.reader(f) reads it
# back, automatically splitting each line into a list of values -
# handling the comma-parsing for us instead of manual string splitting.
```

## File Handling: 4. JSON Files

### Explanation
JSON (JavaScript Object Notation) is a text format for storing 
structured data as key-value pairs, very similar to a Python 
dictionary. Python's json module converts between Python objects 
and JSON text automatically.

```python
import json

student = {"name": "Balaji", "age": 24, "course": "AI/ML"}

# Writing a Python dictionary as JSON
with open("student.json", "w") as f:
    json.dump(student, f)

# Reading JSON back into a Python dictionary
with open("student.json", "r") as f:
    loaded_data = json.load(f)
print(loaded_data)
print(type(loaded_data))

# Where it's used: json.dump(student, f) converts the Python
# dictionary into JSON text format and writes it to the file.
# json.load(f) reads that JSON text back and automatically converts
# it back into a real Python dictionary - confirmed by type() showing <class 'dict'>.
```

## File Handling: 5. Pickle Files

### Explanation
Pickling is Python's way of saving ANY Python object (not just 
text/dictionaries) directly to a file in binary form - lists, 
custom class objects, whatever - and loading it back exactly as 
it was, without converting it to text first like JSON does.

```python
import pickle

data = {"name": "Balaji", "scores": [95, 88, 92], "passed": True}

# Writing (pickling) the object to a file - note "wb" = write binary
with open("data.pkl", "wb") as f:
    pickle.dump(data, f)

# Reading (unpickling) it back - note "rb" = read binary
with open("data.pkl", "rb") as f:
    loaded_data = pickle.load(f)
print(loaded_data)

# Where it's used: pickle.dump(data, f) saves the Python object
# in binary form (not readable text like JSON). pickle.load(f)
# reads that binary data back into an identical Python object -
# "wb"/"rb" (binary modes) are required for pickle, unlike the
# plain text mode used for .txt/.csv/.json files.
```

## File Handling: 6. Context Manager (with)

### Explanation
The 'with' statement automatically handles opening AND closing a 
file (or other resource) for you - even if an error occurs inside 
the block. Without 'with', you'd have to manually call 
file.close() every time, and forgetting it can cause file 
corruption or resource leaks.

```python
# Without context manager - manual close required
f = open("manual.txt", "w")
f.write("Written without context manager")
f.close()   # must remember to do this yourself

# With context manager - closes automatically, even if an error happens
with open("auto.txt", "w") as f:
    f.write("Written with context manager")
# file is already closed here automatically, no need to call f.close()

print("manual.txt closed:", f.closed)

# Where it's used: 'with open(...) as f:' is the context manager.
# The moment the indented block under it ends, Python automatically
# calls f.close() for you - you can verify this by checking f.closed,
# which becomes True the instant the 'with' block finishes, with no
# manual close() call needed.
```

## File Handling: 7. pathlib

### Explanation
pathlib is a modern, object-oriented way to work with file paths 
(checking if a file exists, joining paths, getting file extensions, 
etc.) - a cleaner alternative to manually building path strings.

```python
from pathlib import Path

file_path = Path("sample.txt")

print("Exists?", file_path.exists())
print("File name:", file_path.name)
print("File extension:", file_path.suffix)
print("Full absolute path:", file_path.absolute())

# Where it's used: Path("sample.txt") creates a Path OBJECT
# (not just a plain string). Its built-in properties like .exists(),
# .name, and .suffix let us inspect the file without manually
# writing string-splitting logic ourselves.
```

## File Handling: 8. os

### Explanation
The os module gives access to operating-system-level operations - 
like listing files in a folder, checking the current working 
directory, creating/removing folders, and interacting with the 
file system more broadly than just single files.

```python
import os

print("Current working directory:", os.getcwd())
print("Files in this folder:", os.listdir("."))

os.makedirs("test_folder", exist_ok=True)
print("test_folder created?", os.path.exists("test_folder"))

# Where it's used: os.getcwd() asks the OPERATING SYSTEM for the
# current folder location. os.listdir(".") asks it to list files in
# the current directory. os.makedirs(...) actually creates a real
# new folder on disk - all OS-level actions, not just file content operations.
```

## File Handling: 9. shutil

### Explanation
shutil (shell utilities) handles higher-level file OPERATIONS like 
copying, moving, and deleting entire files or folders - things 
that would take multiple manual steps using just open()/os alone.

```python
import shutil

# Copy a file
shutil.copy("sample.txt", "sample_copy.txt")
print("Copy created?", os.path.exists("sample_copy.txt"))

# Move a file into the test_folder we created earlier
shutil.move("sample_copy.txt", "test_folder/sample_copy.txt")
print("Moved into test_folder?", os.path.exists("test_folder/sample_copy.txt"))

# Where it's used: shutil.copy(...) duplicates a file in one line
# (would need manual read+write with plain file handling). 
# shutil.move(...) relocates a file to a different folder - both
# are higher-level operations shutil provides beyond basic open()/os.
```

## File Handling: 10. File Validation

### Explanation
File validation means checking whether a file exists, is the 
right type, or is accessible BEFORE trying to read/write it - 
preventing crashes from missing files or wrong formats.

```python
from pathlib import Path

def read_file_safely(filename):
    file_path = Path(filename)
    
    if not file_path.exists():
        print(f"Error: {filename} does not exist!")
        return
    
    if file_path.suffix != ".txt":
        print(f"Error: {filename} is not a .txt file!")
        return
    
    with open(file_path, "r") as f:
        print(f.read())

read_file_safely("sample.txt")        # valid - exists and is .txt
read_file_safely("missing.txt")       # invalid - doesn't exist
read_file_safely("student.json")      # invalid - wrong extension

# Where it's used: 'if not file_path.exists():' and 
# 'if file_path.suffix != ".txt":' are the validation checks -
# they run BEFORE attempting to open the file, catching problems
# early instead of letting the program crash with an unhandled error.
```

## File Handling: 11. Directory Operations

### Explanation
Directory operations involve creating, listing, navigating, and 
removing folders (not individual files) - managing the folder 
structure itself.

```python
import os

# Create a directory
os.makedirs("project_data/logs", exist_ok=True)
print("Nested folders created?", os.path.exists("project_data/logs"))

# List contents of a directory
print("Contents of project_data:", os.listdir("project_data"))

# Remove an empty directory
os.rmdir("project_data/logs")
print("logs folder removed?", not os.path.exists("project_data/logs"))

# Where it's used: os.makedirs(...) creates NESTED folders in one
# call (project_data, then logs inside it). os.listdir(...) shows
# what's inside a directory. os.rmdir(...) removes an empty
# directory - all operating on FOLDERS themselves, not file content.
```
