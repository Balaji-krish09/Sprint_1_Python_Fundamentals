## OOP: 1 & 2. Class and Object

### What is a Class?
A class is a blueprint or plan. It defines what something should 
look like and what it can do — but it's not a real, usable thing 
on its own. Just the design.

### What is an Object?
An object is the actual real thing created from that blueprint. 
You can create many objects from a single class, and each one 
exists independently of the others.

### Real-Life Analogy: Cookie Cutter
Think of a cookie cutter — it decides the shape of the cookie, but 
you can't eat the cutter itself. The class is the cutter. Every 
actual cookie you cut out using it is an object — real, edible, 
and separate from the others, even though they all came from the 
same cutter.

```python
class Student:      # the blueprint / cookie cutter
    pass

# Creating actual objects (cookies) from the class
student1 = Student()
student2 = Student()

print(type(student1))         # confirms student1 is a Student object
print(student1 is student2)   # False - two separate, independent objects
```

## OOP: 3 & 4. Constructor (__init__) and Instance Variables

### What is a Constructor?
A special method (`__init__`) that runs automatically the moment 
an object is created. It's used to set up the object's starting 
data.

### What are Instance Variables?
Data that belongs to one specific object only. Different objects 
of the same class can hold completely different values for the 
same variable.

### Real-Life Analogy: Decorating a Cookie
The moment a cookie comes out of the cutter, it gets decorated 
right away (constructor runs automatically). Each cookie can get 
different sprinkles/decoration (instance variables) — one cookie's 
decoration doesn't affect another's.

```python
class Student:
    def __init__(self, name, age):   # constructor - runs automatically
        self.name = name             # instance variable
        self.age = age               # instance variable

# Creating two objects - each gets its OWN data
student1 = Student("Balaji", 24)
student2 = Student("priya", 22)

print(student1.name, student1.age)   # Balaji 24
print(student2.name, student2.age)   # Priya 22 - different data, same class
```

## OOP: 5-8. Class Variables, Instance Methods, Static Methods, Class Methods

### Class Variable
Data shared by ALL objects of a class - one single copy, not one 
per object.

### Instance Method
A normal method that works using a specific object's own data 
(uses `self`).

### Static Method
A method that doesn't need object data or class data - just 
grouped inside the class for organization (uses `@staticmethod`).

### Class Method
A method that works with the class itself, not one object 
(uses `cls` and `@classmethod`).

### Real-Life Analogy: School
- Class Variable = the school's name (shared by every student)
- Instance Method = a student telling you THEIR name (depends on the student)
- Static Method = a general school rule, like "starts at 9 AM" - independent of any student
- Class Method = renaming the school itself - affects everyone

```python
class Student:
    school_name = "Stackly Academy"   # class variable - shared by all

    def __init__(self, name):
        self.name = name              # instance variable - unique per object

    def introduce(self):              # instance method - uses self
        print(f"Hi, I'm {self.name} from {self.school_name}")

    @staticmethod
    def school_timing():              # static method - no self, no cls
        print("School starts at 9 AM")

    @classmethod
    def rename_school(cls, new_name): # class method - uses cls
        cls.school_name = new_name

s1 = Student("Balaji")
s2 = Student("Priya")

s1.introduce()
s2.introduce()

Student.school_timing()

Student.rename_school("New Stackly Academy")
s1.introduce()   # both students now show updated school name
s2.introduce()
```

## OOP: 9 & 10. Properties (@property) and Encapsulation

### What is Encapsulation?
Hiding and protecting an object's internal data so it can't be 
changed carelessly from outside. Access is controlled through 
defined methods instead of direct access.

### What is a Property (@property)?
A way to make a method behave like a normal variable when read or 
set, while still allowing validation/control to run behind the 
scenes.

### Real-Life Analogy: ATM Machine
You never touch the bank's actual money storage directly - you go 
through a controlled interface (card, PIN, request). The machine 
checks things before allowing access. A @property is like a 
"balance inquiry" that looks like a simple number, but is secretly 
running checks behind the scenes.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance   # double underscore = "private", hidden from outside

    @property
    def balance(self):             # looks like a variable, but it's a method
        return self.__balance

    @balance.setter
    def balance(self, amount):
        if amount < 0:              # validation/control before changing data
            print("Balance cannot be negative!")
        else:
            self.__balance = amount

account = BankAccount(1000)

print(account.balance)      # reads like a variable, actually calls the property method
account.balance = 2000      # sets like a variable, actually calls the setter
print(account.balance)

account.balance = -500      # blocked by validation
print(account.balance)      # still 2000, unchanged
```

## OOP: 11 & 12. Abstraction and Inheritance

### What is Abstraction?
Showing only the important, necessary details to the user while 
hiding the complicated internal implementation.

### What is Inheritance?
A class (child) reuses and extends the code of another class 
(parent), instead of rewriting everything from scratch.

### Real-Life Analogy
Abstraction: Driving a car - you use the steering wheel and pedals 
without needing to know how the engine internally works.
Inheritance: A Car is a type of Vehicle. Instead of redefining 
"has wheels, can move" every time, Car inherits that from Vehicle, 
and only adds what's specifically different about cars.

```python
# Parent class
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

    def general_info(self):
        print(f"This is a {self.brand} vehicle. It can move.")

# Child class - inherits from Vehicle
class Car(Vehicle):
    def __init__(self, brand, doors):
        super().__init__(brand)   # reuse parent's constructor
        self.doors = doors

    def car_info(self):
        print(f"This car has {self.doors} doors.")

my_car = Car("Toyota", 4)
my_car.general_info()   # inherited from Vehicle
my_car.car_info()       # defined in Car itself
```

## OOP: 13-15. Multiple, Multilevel, and Hierarchical Inheritance

### Multiple Inheritance
One child class inherits from two or more parent classes at once.
Analogy: A kid inherits traits from BOTH Mom and Dad.

### Multilevel Inheritance
A chain of inheritance - grandparent to parent to child, each 
level building on the one above.
Analogy: Grandfather -> Father -> Son, traits pass down the chain.

### Hierarchical Inheritance
One single parent class, with multiple different child classes 
inheriting from it separately.
Analogy: One Vehicle class, but Car, Bike, and Truck all inherit 
from it independently.

```python
# --- Multiple Inheritance ---
class Mother:
    def eye_color(self):
        print("Eyes: Brown (from Mother)")

class Father:
    def height(self):
        print("Height: Tall (from Father)")

class Child(Mother, Father):   # inherits from BOTH
    pass

c = Child()
c.eye_color()
c.height()

print("\n--- Multilevel Inheritance ---")
class Grandfather:
    def family_trait(self):
        print("Family Trait: Hardworking")

class Father2(Grandfather):     # inherits from Grandfather
    pass

class Son(Father2):             # inherits from Father2 (which has Grandfather's trait)
    pass

s = Son()
s.family_trait()   # comes from all the way up the chain

print("\n--- Hierarchical Inheritance ---")
class Vehicle:
    def move(self):
        print("This vehicle can move.")

class Car(Vehicle):    # child 1
    pass

class Bike(Vehicle):   # child 2 - separate from Car, same parent
    pass

Car().move()
Bike().move()
```

## OOP: 16-18. Polymorphism, Method Overriding, Method Overloading

### What is Polymorphism?
The same method name behaves differently depending on which object 
calls it.

### What is Method Overriding?
A child class redefines a method it inherited from its parent, 
giving it new/different behavior.

### What is Method Overloading (Python Approach)?
Python doesn't support true overloading like some other languages. 
Instead, default arguments are used so one method can handle 
different numbers of inputs.

### Real-Life Analogy
Polymorphism: Pressing "Start" on a car vs. a washing machine - 
same action, different result depending on the object.
Overriding: Animal has make_sound(), but Dog overrides it to bark 
and Cat overrides it to meow.
Overloading: A pizza order form that works whether you specify 1 
topping or 3, using default values.

```python
# --- Polymorphism + Method Overriding ---
class Animal:
    def make_sound(self):
        print("Some generic animal sound")

class Dog(Animal):
    def make_sound(self):        # overriding the parent's method
        print("Bark!")

class Cat(Animal):
    def make_sound(self):        # overriding the parent's method
        print("Meow!")

animals = [Dog(), Cat(), Animal()]
for a in animals:
    a.make_sound()   # polymorphism - same method call, different results

print("\n--- Method Overloading (Python style) ---")
class Pizza:
    def order(self, topping1, topping2=None, topping3=None):
        toppings = [t for t in [topping1, topping2, topping3] if t]
        print("Toppings:", toppings)

p = Pizza()
p.order("Cheese")                          # 1 topping
p.order("Cheese", "Mushroom")              # 2 toppings
p.order("Cheese", "Mushroom", "Olives")    # 3 toppings
```

## OOP: 19-21. Composition, Aggregation, Association

### Composition (strongest - "owns")
One object owns another. If the owner is destroyed, the owned 
part is destroyed too - it cannot exist independently.
Analogy: A House and its Rooms. Demolish the house, the rooms are gone too.

### Aggregation (weaker - "has, but independent")
One object has/uses another, but both can exist independently.
Analogy: A Teacher and a School. If the school shuts down, the 
teacher still exists and can work elsewhere.

### Association (loosest - "just interacts")
Two objects simply know about or interact with each other, with 
no ownership at all.
Analogy: A Student and a Library Book. They interact temporarily, 
neither depends on the other.

```python
# --- Composition ---
class Room:
    def __init__(self, name):
        self.name = name

class House:
    def __init__(self):
        self.rooms = [Room("Bedroom"), Room("Kitchen")]  # House CREATES its own rooms

house = House()
print("Composition - Rooms in house:", [r.name for r in house.rooms])
# If 'house' is deleted, its rooms are deleted too - they don't exist independently

# --- Aggregation ---
class Teacher:
    def __init__(self, name):
        self.name = name

class School:
    def __init__(self, teacher):        # School RECEIVES an already-existing teacher
        self.teacher = teacher

teacher1 = Teacher("Mr. Kumar")
school = School(teacher1)
print("Aggregation - Teacher:", school.teacher.name)
# teacher1 exists independently - even without 'school', teacher1 is still usable

# --- Association ---
class Student:
    def __init__(self, name):
        self.name = name

class Book:
    def __init__(self, title):
        self.title = title

def borrow_book(student, book):    # just a temporary interaction, no ownership
    print(f"{student.name} borrowed '{book.title}'")

s1 = Student("Balaji")
b1 = Book("Python Basics")
borrow_book(s1, b1)   # they interact, but neither owns the other
```

## OOP: 22-26. Magic Methods, super(), __str__, __repr__, __len__

### What are Magic (Dunder) Methods?
Special methods with double underscores (like __init__) that 
Python calls AUTOMATICALLY in certain situations - not called 
directly by us.

### What is super()?
Lets a child class call its parent class's version of a method, 
instead of rewriting that logic.

### __str__() 
Controls what's shown when you print() an object - a readable, 
human-friendly version.

### __repr__()
Controls the official/technical representation of an object - 
mainly used for debugging.

### __len__()
Lets Python's built-in len() function work on your own custom objects.

### Real-Life Analogy: Doorbell
Dunder methods are automatic reactions - pressing a doorbell 
(action) automatically triggers a chime (__init__ triggers on 
creation, __str__ triggers on print). You don't manually ring it - 
the action triggers it automatically.

```python
class Book:
    def __init__(self, title, pages):
        self.title = title
        self.pages = pages

    def __str__(self):                    # triggered automatically by print()
        return f"Book: {self.title}"

    def __repr__(self):                   # triggered when you just type the object name
        return f"Book(title={self.title!r}, pages={self.pages})"

    def __len__(self):                    # triggered automatically by len()
        return self.pages

class Ebook(Book):
    def __init__(self, title, pages, file_size):
        super().__init__(title, pages)    # reuse parent's constructor
        self.file_size = file_size

    def __str__(self):
        base = super().__str__()          # reuse parent's __str__, then add to it
        return f"{base} (Ebook, {self.file_size}MB)"

b = Book("Python Basics", 250)
print(b)          # triggers __str__ automatically
print(repr(b))    # triggers __repr__ automatically
print(len(b))     # triggers __len__ automatically

e = Ebook("AI Fundamentals", 300, 15)
print(e)          # uses Ebook's __str__, which calls Book's __str__ via super()
```

## OOP: 27 & 28 (Bonus). Dataclasses and Singleton Pattern

### What is a Dataclass?
A shortcut (@dataclass) that auto-generates boilerplate code like 
__init__ and __repr__ for classes that mainly just store data - 
saves writing repetitive code manually.

### What is the Singleton Pattern?
A design where a class is restricted to only ever having ONE 
object, no matter how many times someone tries to create a new one.

### Real-Life Analogy
Dataclass: Filling a pre-printed form instead of writing the whole 
structure by hand each time.
Singleton: A country's president - no matter how many times you 
try to "create" one, there's only ever one active at a time.

```python
from dataclasses import dataclass

# --- Dataclass ---
@dataclass
class Product:
    name: str
    price: float
    # __init__ and __repr__ are auto-generated - no need to write them manually

p = Product("Laptop", 55000.0)
print(p)   # auto-generated __repr__ gives readable output

print("\n--- Singleton Pattern ---")
class Singleton:
    _instance = None   # keeps track of the one and only instance

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)   # create only if it doesn't exist yet
        return cls._instance                        # otherwise, return the existing one

s1 = Singleton()
s2 = Singleton()
print("s1 is s2:", s1 is s2)   # True - both point to the SAME single object
```
