## 1. Variables and Data Types

### Definition
A **variable** is a named place to store a value. It is like a box with a label that keeps information. The value inside the variable can change whenever we want.

A **data type** tells Python what kind of information is stored in a variable:
* **Integer (`int`)**: Whole numbers (e.g., `24`)
* **Float (`float`)**: Decimal numbers (e.g., `5.9`)
* **String (`str`)**: Text enclosed in quotes (e.g., `"Balaji"`)
* **Boolean (`bool`)**: `True` or `False`

### Real-Life Example
Think of labeled storage containers in a kitchen:
* Container labeled **"Age"** holds the number `24`.
* Container labeled **"Name"** holds text `"Balaji"`.
* We can use the label **"Age"** anywhere in our program instead of writing `24` every time.

```python
# ==========================================
# 1. Variables and Data Types
# ==========================================

# Variable declarations
age = 24                   # Integer (int)
height = 5.9               # Float (float)
name = "Balaji"            # String (str)
is_student = True          # Boolean (bool)

# Displaying values and their data types
print("Name:", name, "| Data Type:", type(name))
print("Age:", age, "| Data Type:", type(age))
print("Height:", height, "| Data Type:", type(height))
print("Is Student:", is_student, "| Data Type:", type(is_student))
```

## 2. Type Casting and Operators

### Definition
* **Type Casting**: Changing one data type into another (e.g., converting text `"25"` into a usable number `25`). Both represent twenty-five, but Python treats text and numbers completely differently in memory.
* **Operators**: Symbols used to perform calculations, comparisons, or logical checks.
  * **Arithmetic Operators**: `+` (Add), `-` (Subtract), `*` (Multiply), `/` (Divide), `//` (Floor Divide), `%` (Modulus/Remainder)
  * **Comparison Operators**: `>` (Greater than), `<` (Less than), `==` (Equal to), `!=` (Not equal to)

### Real-Life Example
Think of receiving an SMS alert with an amount like `"500"`. To Python, `"500"` is just text (a string). Before a banking app can add it to your total account balance, it must perform **Type Casting** to convert `"500"` into an integer `500`.

```python
# ==========================================
# 2. Type Casting and Operators
# ==========================================

# Type Casting Example
raw_amount = "500"                  # String (text)
converted_amount = int(raw_amount)  # Explicitly cast to Integer

bonus = 50
total_balance = converted_amount + bonus  # Arithmetic addition

print("Raw String:", raw_amount, "| Type:", type(raw_amount))
print("Converted Int:", converted_amount, "| Type:", type(converted_amount))
print("Total Balance after addition:", total_balance)

print("\n--- Operators Check ---")
num1 = 10
num2 = 3

print("Addition (10 + 3):", num1 + num2)
print("Standard Division (10 / 3):", num1 / num2)      # Returns float
print("Floor Division (10 // 3):", num1 // num2)       # Drops decimals
print("Modulus/Remainder (10 % 3):", num1 % num2)      # Returns remainder
print("Is Balance > 400?:", total_balance > 400)       # Comparison operator (Boolean result)
```

## 3. Input, Output, and Comments

### Definition
* **Input**: Giving information to the program (e.g., using `input()`). Note that `input()` *always* returns data as a **String (`str`)**.
* **Output**: Displaying the result back to the user (e.g., using `print()`).
* **Comments**: Notes written inside the code for human readability. Python completely ignores comments when running the program.
  * **Single-line comment**: Uses the `#` symbol.
  * **Multi-line / Docstring**: Uses triple quotes (`""" ... """`).

### Real-Life Example
Think of a simple calculator:
* **Input**: You type `5` and `3` on the keypad.
* **Process**: The calculator adds them together (`5 + 3`).
* **Output**: The screen displays `8`.

```python
# ==========================================
# 3. Input, Output, and Comments
# ==========================================

# Single-line comment: Setting default values for calculation
a = 5
b = 3

"""
Docstring / Multi-line Comment:
This section processes the addition of two numbers 
and displays the formatted output to the console.
"""

# Calculating addition
result = a + b

# Output
print("Input A:", a)
print("Input B:", b)
print("Output (Sum):", result)
```
