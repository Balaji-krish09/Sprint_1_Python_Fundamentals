## Exception Handling: 1. try and except

### Explanation
try lets you run code that might cause an error, without crashing 
the whole program. If an error occurs inside the try block, Python 
immediately jumps to the matching except block instead of stopping.

```python
try:
    result = 10 / 0
    print(result)
except ZeroDivisionError:
    print("You can't divide by zero!")

# Where it's used: 'result = 10 / 0' causes a ZeroDivisionError inside try.
# Python jumps straight to 'except ZeroDivisionError:' and runs it,
# printing the message. 'print(result)' never runs since the error
# happened before reaching that line.
```

## Exception Handling: 2. else

### Explanation
The else block runs ONLY if the try block completed with no errors 
at all. Useful for code that should only run when everything in 
try succeeded.

```python
try:
    result = 10 / 2
except ZeroDivisionError:
    print("Cannot divide by zero!")
else:
    print("Success! Result is:", result)

# Where it's used: '10 / 2' causes no error, so 'except' is skipped
# entirely, and Python runs the 'else:' block instead, printing the
# success message. If the division had failed, else would be skipped.
```

## Exception Handling: 3. finally

### Explanation
The finally block ALWAYS runs, no matter what - whether an error 
happened or not. Typically used for cleanup actions (like closing 
a file) that must happen regardless of success or failure.

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero!")
finally:
    print("This always runs, no matter what.")

# Where it's used: 'finally:' runs after 'except' handles the error -
# but it would run just as certainly even with no error at all, or
# even if the error wasn't caught. It's guaranteed to execute.
```

## Exception Handling: 4. raise

### Explanation
raise lets you manually trigger an exception yourself, on purpose, 
even if Python wouldn't normally throw an error there. Useful when 
you want to enforce a rule in your own code.

```python
def set_age(age):
    if age < 0:
        raise ValueError("Age cannot be negative!")
    print("Age set to:", age)

try:
    set_age(-5)
except ValueError as e:
    print("Caught an error:", e)

# Where it's used: 'raise ValueError(...)' manually triggers an
# exception when age is negative, even though Python itself would
# never automatically flag a negative number as an error. The raise
# happens inside set_age(), and the try/except around the function
# call catches and handles it.
```

## Exception Handling: 5. assert

### Explanation
assert checks whether a condition is True. If it's False, Python 
automatically raises an AssertionError. Commonly used to catch bugs 
early by verifying assumptions while writing/testing code.

```python
def divide(a, b):
    assert b != 0, "b cannot be zero!"
    return a / b

try:
    print(divide(10, 0))
except AssertionError as e:
    print("Assertion failed:", e)

# Where it's used: 'assert b != 0, "..."' checks the condition
# b != 0. Since b is 0, the condition is False, so Python
# automatically raises an AssertionError with that message -
# caught by the except block below.
```

## Exception Handling: 6. Custom Exceptions

### Explanation
A custom exception is your own exception class, created by 
inheriting from Python's built-in Exception class. Useful when 
built-in exceptions don't clearly describe the specific error in 
your own program's logic.

```python
class InsufficientBalanceError(Exception):   # custom exception class
    pass

def withdraw(balance, amount):
    if amount > balance:
        raise InsufficientBalanceError("Not enough balance for this withdrawal!")
    return balance - amount

try:
    withdraw(1000, 5000)
except InsufficientBalanceError as e:
    print("Custom error caught:", e)

# Where it's used: 'class InsufficientBalanceError(Exception):'
# defines a brand new exception type by inheriting from Exception.
# It's raised manually inside withdraw() using 'raise', and caught
# using 'except InsufficientBalanceError', just like any built-in
# exception would be.
```

## Exception Handling: 7. ValueError

### When it occurs
Raised when a function receives an argument of the correct type, 
but an inappropriate value - e.g., trying to convert a non-numeric 
string into a number.

```python
try:
    age = int("twenty-five")   # "twenty-five" is a string, not a valid number
except ValueError:
    print("Invalid input! Please enter a number.")

# Where it's used: int() expects a string that LOOKS like a number.
# "twenty-five" is a valid string (correct type) but not a valid
# numeric value, so Python raises ValueError - caught by the except block.
```

## Exception Handling: 8. TypeError

### When it occurs
Raised when an operation or function is applied to an object of an 
inappropriate type - e.g., trying to add a string and an integer 
directly.

```python
try:
    result = "5" + 10   # can't combine a string and an integer with +
except TypeError:
    print("Cannot combine these types directly!")

# Where it's used: "5" is a string and 10 is an integer. Python's +
# operator doesn't know how to combine these two different types,
# so it raises TypeError - caught by the except block.
```

## Exception Handling: 9. IndexError

### When it occurs
Raised when you try to access a list/sequence position that 
doesn't exist - e.g., asking for index 5 in a list that only has 
3 items.

```python
numbers = [10, 20, 30]
try:
    print(numbers[5])   # only indexes 0, 1, 2 exist
except IndexError:
    print("That index doesn't exist in the list!")

# Where it's used: the list 'numbers' only has valid indexes 0, 1, 2.
# Asking for numbers[5] goes out of range, so Python raises
# IndexError - caught by the except block.
```

## Exception Handling: 10. KeyError

### When it occurs
Raised when you try to access a dictionary key that doesn't exist.

```python
student = {"name": "Balaji", "age": 24}
try:
    print(student["marks"])   # "marks" key doesn't exist in this dictionary
except KeyError:
    print("That key doesn't exist in the dictionary!")

# Where it's used: the dictionary 'student' only has keys "name" and
# "age". Asking for student["marks"] looks for a key that was never
# defined, so Python raises KeyError - caught by the except block.
```

## Exception Handling: 11. AttributeError

### When it occurs
Raised when you try to access a method or attribute that doesn't 
exist on an object - e.g., calling a list method on an integer.

```python
number = 25
try:
    number.append(5)   # append() is a list method, not available on integers
except AttributeError:
    print("This object doesn't have that method!")

# Where it's used: 'number' is an integer, and integers don't have
# an .append() method - that only exists on lists. Python raises
# AttributeError - caught by the except block.
```

## Exception Handling: 12. NameError

### When it occurs
Raised when you try to use a variable or function name that hasn't 
been defined anywhere in the current scope.

```python
try:
    print(total_price)   # 'total_price' was never defined anywhere
except NameError:
    print("That variable hasn't been defined yet!")

# Where it's used: 'total_price' is used inside print() but was
# never assigned a value anywhere before this line, so Python
# raises NameError - caught by the except block.
```

## Exception Handling: 13. FileNotFoundError

### When it occurs
Raised when you try to open a file that doesn't exist at the given 
path.

```python
try:
    with open("nonexistent_file.txt", "r") as f:
        content = f.read()
except FileNotFoundError:
    print("That file doesn't exist!")

# Where it's used: "nonexistent_file.txt" doesn't actually exist in
# the current folder. open() tries to locate it and fails, so
# Python raises FileNotFoundError - caught by the except block.
```

## Exception Handling: 14. ZeroDivisionError

### When it occurs
Raised when you try to divide a number by zero, which is 
mathematically undefined.

```python
try:
    result = 10 / 0
except ZeroDivisionError:
    print("Cannot divide by zero!")

# Where it's used: '10 / 0' attempts division by zero, which is
# mathematically undefined, so Python raises ZeroDivisionError -
# caught by the except block.
```

## Exception Handling: 15. ImportError

### When it occurs
Raised when a specific name (function, class, variable) cannot be 
imported from a module, even though the module itself exists.

```python
try:
    from math import non_existent_function   # math module exists, but this function doesn't
except ImportError:
    print("That specific function doesn't exist in this module!")

# Where it's used: 'math' is a real, existing module, but it has no
# function called 'non_existent_function'. Python raises
# ImportError - caught by the except block.
```

## Exception Handling: 16. ModuleNotFoundError

### When it occurs
Raised when the module itself doesn't exist at all (a more specific 
version of ImportError, just for missing modules).

```python
try:
    import some_fake_module_xyz   # this module was never installed/doesn't exist
except ModuleNotFoundError:
    print("That module doesn't exist!")

# Where it's used: 'some_fake_module_xyz' isn't a real installed
# Python module. Python can't find it at all, so it raises
# ModuleNotFoundError - caught by the except block, which is a
# more specific subclass of ImportError.
```
